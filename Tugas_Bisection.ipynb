```python
# Cell 1: Import library yang dibutuhkan
import numpy as np
import matplotlib.pyplot as plt
print("Library siap!")
```

```python
# Cell 2: Mendefinisikan fungsi metode Bisection
def metode_bisection(f, a, b, tol=1e-6, max_iter=50):
    # Cek validitas interval (nilai f(a) dan f(b) harus beda tanda)
    if f(a) * f(b) >= 0:
        print("Error: f(a) dan f(b) harus memiliki tanda berlawanan.")
        return None, 0

    iterasi = 0
    # Lakukan perulangan hingga batas maksimal
    for i in range(max_iter):
        iterasi += 1
        c = (a + b) / 2.0 # Menghitung titik tengah

        # Kriteria berhenti jika error sangat kecil (mendekati 0)
        if abs(f(c)) < tol or (b - a) / 2.0 < tol:
            return c, iterasi

        # Menentukan interval baru yang mengandung akar
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return c, iterasi
```

```python
# Cell 3: Uji coba persamaan f(x) = x^3 - x - 2
def f(x):
    return x**3 - x - 2

# Membuat plot grafik untuk melihat letak akar persamaan
x_vals = np.linspace(0, 3, 100)
y_vals = f(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, label="f(x) = x^3 - x - 2")
plt.axhline(0, color='red', linestyle='--') # Garis bantu sumbu X
plt.axvline(0, color='red', linestyle='--') # Garis bantu sumbu Y
plt.grid(True)
plt.legend()
plt.title("Grafik Fungsi Non-Linear")
plt.show()

# Menjalankan metode Bisection di interval [1, 2] karena f(1) negatif dan f(2) positif
akar, total_iterasi = metode_bisection(f, 1, 2, tol=1e-5)

if akar is not None:
    print(f"\nAkar yang ditemukan: x = {akar:.6f}")
    print(f"Total iterasi: {total_iterasi}")
    print(f"Verifikasi f(x): {f(akar):.6e}")
```
