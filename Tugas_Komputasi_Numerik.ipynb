```python
# Cell 1: Import library numpy untuk operasi matriks
import numpy as np
print("Library berhasil di-import!")
```

```python
# Cell 2: Mendefinisikan fungsi Eliminasi Gauss dengan Partial Pivoting
def eliminasi_gauss(A, b):
    n = len(b)
    # Menggabungkan matriks A dan vektor b menjadi matriks augmented (3x4)
    aug = np.hstack([A, b.reshape(-1, 1)]).astype(float)

    # Tahap Eliminasi Maju
    for i in range(n - 1):
        # Mencari nilai maksimum absolut di kolom i untuk Partial Pivoting
        max_row = i + np.argmax(np.abs(aug[i:, i]))

        # Jika nilai maksimumnya 0, berarti sistem tidak punya solusi unik
        if abs(aug[max_row, i]) < 1e-12:
            return None

        # Menukar baris jika baris dengan nilai maksimum bukan di posisi i
        if max_row != i:
            aug[[i, max_row]] = aug[[max_row, i]]

        # Proses eliminasi elemen di bawah diagonal utama
        for j in range(i + 1, n):
            faktor = aug[j, i] / aug[i, i]
            aug[j, i:] -= faktor * aug[i, i:]

    # Tahap Substitusi Balik (Mencari nilai x dari bawah ke atas)
    x = np.zeros(n)
    x[-1] = aug[-1, -1] / aug[-1, -2]
    for i in range(n - 2, -1, -1):
        jumlah = aug[i, -1] - np.dot(aug[i, i+1:n], x[i+1:])
        x[i] = jumlah / aug[i, i]

    return x
```

```python
# Cell 3: Contoh soal dan memanggil fungsi
# Sistem persamaan yang digunakan:
# 3x + 2y - z = 10
# 2x - y + 3z = 8
# x + 4y + 2z = 15

A = np.array([[ 3,  2, -1],
              [ 2, -1,  3],
              [ 1,  4,  2]])
b = np.array([10, 8, 15])

print("Matriks Koefisien (A):")
print(A)
print("\nVektor Konstanta (b):", b)

# Memanggil fungsi yang sudah dibuat di Cell 2
solusi = eliminasi_gauss(A, b)

# Menampilkan hasil
if solusi is not None:
    print("\n--- Hasil Solusi ---")
    print(f"x = {solusi[0]:.4f}")
    print(f"y = {solusi[1]:.4f}")
    print(f"z = {solusi[2]:.4f}")
```
